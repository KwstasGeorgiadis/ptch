```python
import requests
import json
from fastapi import FastAPI

url = 'https://www.thesportsdb.com/api/v1/json/3/search_all_teams.php'
response = requests.get(url, params = {'l': 'English Premier League'})
data = response.json()
print(json.dumps(data, indent=2))
teams = data['teams']
#print(json.dumps(teams, indent=2))

for team in teams:
    team_name = team['strTeam']
    team_id = team['idTeam']
    print(f'Team: {team_name}, ID: {team_id}')





from fastapi import FastAPI

app = FastAPI()

teams = [
    {"id": 1, "name": "Atlanta Hawks", "short": "ATL"},
    {"id": 2, "name": "Boston Celtics", "short": "BOS"},
]

@app.get("/teams")
def get_teams():
    return teams

@app.get("/teams/{team_id}")
def get_team(team_id: int):
    for team in teams:
        if team["id"] == team_id:
            return team
    return {"error": "not found"}

```
